```python
from pathlib import Path

def save_clean_df(df, out_path: str):
    out_path = Path(out_path)

    # Try parquet first
    try:
        df.to_parquet(out_path, index=False, engine="pyarrow")
        print(f"Saved parquet: {out_path}")
        return
    except Exception as e:
        print(f"Parquet failed: {type(e).__name__}: {e}")

    # Fallback 1: fastparquet, if installed
    try:
        alt = out_path.with_suffix(".fastparquet.parquet")
        df.to_parquet(alt, index=False, engine="fastparquet")
        print(f"Saved parquet with fastparquet: {alt}")
        return
    except Exception as e:
        print(f"fastparquet failed: {type(e).__name__}: {e}")

    # Fallback 2: CSV
    csv_path = out_path.with_suffix(".csv")
    df.to_csv(csv_path, index=False)
    print(f"Saved CSV fallback: {csv_path}")
```

```python
from __future__ import annotations

import re
from pathlib import Path

import numpy as np
import pandas as pd
import openpyxl


FILE = Path("undesa_pd_2024_ims_stock_by_sex_destination_and_origin.xlsx")
NA_TOKENS = {"..", "—", "~"}


def clean_text(x):
    """Trim whitespace, collapse spaces, remove footnote asterisks."""
    if pd.isna(x):
        return pd.NA
    s = str(x).replace("\xa0", " ").strip()
    s = re.sub(r"\s+", " ", s)
    return s.rstrip("*")


def parse_year_from_title(text: str) -> int:
    m = re.search(r"(\d{4})\s*$", str(text))
    if not m:
        raise ValueError(f"Could not parse year from: {text!r}")
    return int(m.group(1))


def parse_table1(ws) -> pd.DataFrame:
    """
    Table 1 layout:
    row 11 = header row
    rows 12:end = data
    columns:
      A:G  = metadata
      H:O  = both sexes (1990..2024)
      P:W  = males
      X:AE = females
    """
    rows = list(ws.iter_rows(min_row=12, max_row=ws.max_row, values_only=True))
    base = pd.DataFrame(rows)

    year_vals = [1990, 1995, 2000, 2005, 2010, 2015, 2020, 2024]
    sex_blocks = [
        ("both_sexes", list(range(7, 15))),  # H:O
        ("male", list(range(15, 23))),        # P:W
        ("female", list(range(23, 31))),      # X:AE
    ]

    parts = []
    for sex, cols in sex_blocks:
        tmp = base.iloc[:, [0, 1, 2, 3, 4, 5, 6] + cols].copy()
        tmp.columns = [
            "row_id",
            "destination",
            "coverage_code",
            "data_type",
            "destination_code",
            "origin",
            "origin_code",
        ] + [str(y) for y in year_vals]
        tmp["sex"] = sex
        parts.append(tmp)

    tidy = pd.concat(parts, ignore_index=True)

    tidy = tidy.melt(
        id_vars=[
            "row_id",
            "destination",
            "coverage_code",
            "data_type",
            "destination_code",
            "origin",
            "origin_code",
            "sex",
        ],
        var_name="year",
        value_name="migrant_stock",
    )

    tidy["destination"] = tidy["destination"].map(clean_text)
    tidy["origin"] = tidy["origin"].map(clean_text)
    tidy["data_type"] = tidy["data_type"].map(clean_text)

    tidy["year"] = tidy["year"].astype(int)
    tidy["coverage_code"] = pd.to_numeric(tidy["coverage_code"], errors="coerce").astype("Int64")
    tidy["destination_code"] = pd.to_numeric(tidy["destination_code"], errors="coerce").astype("Int64")
    tidy["origin_code"] = pd.to_numeric(tidy["origin_code"], errors="coerce").astype("Int64")

    # Robust numeric conversion
    tidy["migrant_stock"] = (
        tidy["migrant_stock"]
        .replace(list(NA_TOKENS), pd.NA)
        .pipe(pd.to_numeric, errors="coerce")
        .astype("Int64")
    )

    # Optional quality flags from the notes sheet
    tidy["flag_B_foreign_born"] = tidy["data_type"].str.contains(r"\bB\b", na=False)
    tidy["flag_C_foreign_citizens"] = tidy["data_type"].str.contains(r"\bC\b", na=False)
    tidy["flag_I_imputed"] = tidy["data_type"].str.contains(r"\bI\b", na=False)
    tidy["flag_R_refugee_adjustment"] = tidy["data_type"].str.contains(r"\bR\b", na=False)

    return tidy


def parse_matrix_sheet(ws, snapshot_year: int) -> pd.DataFrame:
    top = [ws.cell(10, c).value for c in range(1, ws.max_column + 1)]
    sub = [ws.cell(11, c).value for c in range(1, ws.max_column + 1)]
    rows = list(ws.iter_rows(min_row=12, max_row=ws.max_row, values_only=True))
    base = pd.DataFrame(rows)

    dest_meta = base[[0, 1, 3]].copy()
    dest_meta.columns = ["row_id", "destination_region", "destination_code"]

    dest_meta["destination_region"] = dest_meta["destination_region"].map(clean_text)
    dest_meta["destination_code"] = pd.to_numeric(dest_meta["destination_code"], errors="coerce")

    out = []
    for c in range(4, ws.max_column):
        part = dest_meta.copy()

        part["origin_region"] = clean_text(top[c])

        part["origin_code"] = pd.to_numeric(sub[c], errors="coerce")

        part["migrant_stock"] = (
            base.iloc[:, c]
            .replace(list(NA_TOKENS), pd.NA)
            .pipe(pd.to_numeric, errors="coerce")
        )

        part["snapshot_year"] = snapshot_year
        out.append(part)

    result = pd.concat(out, ignore_index=True)

    # enforce dtypes once
    result["destination_code"] = result["destination_code"].astype("Int64")
    result["origin_code"] = result["origin_code"].astype("Int64")
    result["migrant_stock"] = result["migrant_stock"].astype("Int64")

    return result


def load_all(path: Path):
    wb = openpyxl.load_workbook(path, data_only=True)

    table1 = parse_table1(wb["Table 1"])

    matrix_tables = []
    for sheet_name in [f"Table {i}" for i in range(2, 10)]:
        ws = wb[sheet_name]
        year = parse_year_from_title(ws.cell(3, 5).value)
        matrix_tables.append(parse_matrix_sheet(ws, year))

    tables_2_9 = pd.concat(matrix_tables, ignore_index=True)

    return table1, tables_2_9


table1, tables_2_9 = load_all(FILE)

# Optional sanity checks
assert table1.query("destination == 'World' and origin == 'World' and year == 2024 and sex == 'both_sexes'")["migrant_stock"].iat[0] == 304021813
assert (
    table1.query(
        "destination == 'World' and origin == 'World' and year == 2024 and sex == 'both_sexes'"
    )["migrant_stock"].iat[0]
    == 304021813
)
```

```python
table1 = table1.convert_dtypes(dtype_backend="numpy_nullable")
tables_2_9 = tables_2_9.convert_dtypes(dtype_backend="numpy_nullable")

table1.to_parquet("table1.parquet", engine="fastparquet", index=False)
tables_2_9.to_parquet("tables_2_9.parquet", engine="fastparquet", index=False)
```
